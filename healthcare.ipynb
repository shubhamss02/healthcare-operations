```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt

df=pd.read_csv('hospital_er_data.csv')
df.head(10)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ','_')
)
df.info()
```

```python
df['patient_admission_date']=pd.to_datetime(df['patient_admission_date'])
df.info()
```

```python
df['patient_name']=df['patient_first_inital'] + ' ' + df['patient_last_name']
df.head(10)
```

```python
df.duplicated().sum()
```

```python
for col in ['patient_gender', 'patient_race', 'department_referral', 'patient_admission_flag']:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

password = quote_plus("password")

engine = create_engine(
    f"postgresql+psycopg2://postgres:{password}@localhost:5432/healthcare"
)
```

```python
df.to_sql(
    'hospital_er',
    engine,
    if_exists='replace',
    index=False
)
```

```python
pd.read_sql("SELECT * FROM hospital_er LIMIT 5;", engine)
```
